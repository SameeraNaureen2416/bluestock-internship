# Data Analyst Internship – Python Data Analysis
This notebook loads the cleaned sample sales dataset, validates it, calculates KPIs, and creates simple visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
DATA = Path('../01_Week1_Excel/data/sample_sales_cleaned.csv')
df = pd.read_csv(DATA, parse_dates=['order_date'])
df.head()
```

```python
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicate order IDs:', df['order_id'].duplicated().sum())
```

```python
kpis = {
    'Total Sales': df['sales_amount'].sum(),
    'Total Profit': df['profit'].sum(),
    'Orders': df['order_id'].nunique(),
    'Customers': df['customer_id'].nunique(),
    'Average Order Value': df['sales_amount'].mean()
}
pd.Series(kpis)
```

```python
product_summary = df.groupby('product').agg(revenue=('sales_amount','sum'), units=('quantity','sum'), profit=('profit','sum')).sort_values('revenue', ascending=False)
product_summary
```

```python
monthly = df.groupby(df['order_date'].dt.to_period('M'))['sales_amount'].sum()
monthly.plot(kind='line', marker='o', title='Monthly Sales')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
state_summary = df.groupby('state')['sales_amount'].sum().sort_values(ascending=False)
state_summary.plot(kind='bar', title='Sales by State')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```
